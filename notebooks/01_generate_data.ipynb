```python
import pandas as pd
import random
from pathlib import Path
from datetime import datetime, timedelta

# Reproducible results
random.seed(42)

# Find the project folders
PROJECT_DIR = Path.cwd().parent
RAW_DIR = PROJECT_DIR / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)

# 1. PRODUCTS — 30 rows
product_names = [
    ("Wireless Earbuds", "Electronics", "SoundMax", 1499),
    ("Bluetooth Speaker", "Electronics", "SoundMax", 1999),
    ("Smart Watch", "Electronics", "TechPro", 2499),
    ("USB-C Cable", "Electronics", "TechPro", 299),
    ("Power Bank", "Electronics", "VoltX", 999),
    ("Laptop Stand", "Electronics", "WorkEase", 799),
    ("Mechanical Keyboard", "Electronics", "KeyCraft", 2999),
    ("Wireless Mouse", "Electronics", "KeyCraft", 699),
    ("Phone Case", "Accessories", "CoverCo", 399),
    ("Screen Protector", "Accessories", "CoverCo", 199),
    ("Backpack", "Fashion", "UrbanTrail", 1299),
    ("Travel Bottle", "Home", "EcoLife", 499),
    ("Desk Lamp", "Home", "BrightHome", 899),
    ("Water Bottle", "Home", "EcoLife", 349),
    ("Coffee Mug", "Home", "DailyUse", 249),
    ("Running Shoes", "Fashion", "ActiveFit", 2199),
    ("Yoga Mat", "Fitness", "ActiveFit", 899),
    ("Resistance Bands", "Fitness", "FlexGear", 499),
    ("Cotton T-Shirt", "Fashion", "UrbanWear", 599),
    ("Hoodie", "Fashion", "UrbanWear", 1499),
    ("Cookware Set", "Home", "KitchenPro", 2499),
    ("Storage Containers", "Home", "KitchenPro", 699),
    ("Face Moisturizer", "Beauty", "GlowCare", 399),
    ("Sunscreen", "Beauty", "GlowCare", 499),
    ("Shampoo", "Beauty", "PureRoots", 299),
    ("Notebook Set", "Stationery", "PaperNest", 199),
    ("Office Chair", "Furniture", "WorkEase", 5499),
    ("Bedsheet Set", "Home", "SoftLiving", 1299),
    ("Air Fryer", "Appliances", "HomeChef", 3999),
    ("Electric Kettle", "Appliances", "HomeChef", 1199),
]

products = pd.DataFrame(product_names, columns=[
    "product_name", "category", "brand", "unit_price"
])
products.insert(0, "product_id", [f"P{i:03d}" for i in range(1, 31)])

# 2. ORDERS — 500 rows
statuses = ["Delivered", "Delivered", "Delivered", "Cancelled", "Returned"]
start_date = datetime(2025, 1, 1)

orders = []
for i in range(1, 501):
    product = random.choice(products.to_dict("records"))
    order_date = start_date + timedelta(days=random.randint(0, 299))
    quantity = random.randint(1, 4)
    status = random.choice(statuses)

    orders.append({
        "order_id": f"O{i:04d}",
        "product_id": product["product_id"],
        "order_date": order_date.date(),
        "quantity": quantity,
        "unit_price": product["unit_price"],
        "order_status": status,
    })

orders = pd.DataFrame(orders)

# 3. REVIEWS — 250 rows
positive_reviews = [
    "Excellent quality and works perfectly.",
    "Very happy with this purchase.",
    "Good value for money. Would recommend.",
    "Fast delivery and great product.",
    "Exactly as described. Very satisfied.",
]
negative_reviews = [
    "Stopped working after two days.",
    "Poor quality and disappointing experience.",
    "Product arrived damaged.",
    "Battery life is terrible.",
    "Not worth the price. Very disappointed.",
    "The material feels cheap and uncomfortable.",
    "Delivery was late and packaging was damaged.",
]
neutral_reviews = [
    "The product is okay for everyday use.",
    "Average quality for the price.",
    "Works as expected, nothing special.",
    "Decent product but could be improved.",
]

reviews = []
for i in range(1, 251):
    product = random.choice(products.to_dict("records"))
    rating = random.choices(
        [1, 2, 3, 4, 5],
        weights=[8, 12, 20, 30, 30]
    )[0]

    if rating <= 2:
        review_text = random.choice(negative_reviews)
    elif rating == 3:
        review_text = random.choice(neutral_reviews)
    else:
        review_text = random.choice(positive_reviews)

    review_date = start_date + timedelta(days=random.randint(0, 299))

    reviews.append({
        "review_id": f"R{i:04d}",
        "product_id": product["product_id"],
        "review_date": review_date.date(),
        "rating": rating,
        "review_text": review_text,
    })

reviews = pd.DataFrame(reviews)

# Save raw CSV files
products.to_csv(RAW_DIR / "products.csv", index=False)
orders.to_csv(RAW_DIR / "orders.csv", index=False)
reviews.to_csv(RAW_DIR / "reviews.csv", index=False)

print("DATA GENERATION COMPLETE!")
print("-" * 35)
print("Products:", len(products))
print("Orders:", len(orders))
print("Reviews:", len(reviews))
print("Saved to:", RAW_DIR)
```

```python
# Load the generated CSV files
products_check = pd.read_csv(RAW_DIR / "products.csv")
orders_check = pd.read_csv(RAW_DIR / "orders.csv")
reviews_check = pd.read_csv(RAW_DIR / "reviews.csv")

# Check dataset sizes
print("DATASET SIZES")
print("Products:", products_check.shape)
print("Orders:", orders_check.shape)
print("Reviews:", reviews_check.shape)

# Check missing values
print("\nMISSING VALUES")
print("Products:", products_check.isnull().sum().sum())
print("Orders:", orders_check.isnull().sum().sum())
print("Reviews:", reviews_check.isnull().sum().sum())

# Check unique IDs
print("\nDUPLICATE IDs")
print("Products:", products_check["product_id"].duplicated().sum())
print("Orders:", orders_check["order_id"].duplicated().sum())
print("Reviews:", reviews_check["review_id"].duplicated().sum())

# Check product references
valid_product_ids = set(products_check["product_id"])

print("\nINVALID PRODUCT REFERENCES")
print("Orders:", (~orders_check["product_id"].isin(valid_product_ids)).sum())
print("Reviews:", (~reviews_check["product_id"].isin(valid_product_ids)).sum())

# Preview the data
print("\nPRODUCT PREVIEW")
display(products_check.head())

print("\nORDER PREVIEW")
display(orders_check.head())

print("\nREVIEW PREVIEW")
display(reviews_check.head())
```

```python
# 1. Check order prices against the product catalogue
price_check = orders_check.merge(
    products_check[["product_id", "unit_price"]],
    on="product_id",
    how="left",
    suffixes=("_order", "_catalogue")
)

print("ORDER PRICE MISMATCHES:",
      (price_check["unit_price_order"] !=
       price_check["unit_price_catalogue"]).sum())

# 2. Check rating range
print("INVALID RATINGS:",
      (~reviews_check["rating"].between(1, 5)).sum())

# 3. Check allowed order statuses
valid_statuses = {"Delivered", "Cancelled", "Returned"}

print("INVALID ORDER STATUSES:",
      (~orders_check["order_status"].isin(valid_statuses)).sum())

# 4. Check dates
orders_check["order_date"] = pd.to_datetime(
    orders_check["order_date"], errors="coerce"
)
reviews_check["review_date"] = pd.to_datetime(
    reviews_check["review_date"], errors="coerce"
)

print("INVALID ORDER DATES:", orders_check["order_date"].isna().sum())
print("INVALID REVIEW DATES:", reviews_check["review_date"].isna().sum())

print("\nBUSINESS LOGIC CHECK COMPLETE!")
```

```python
# Step 1: Make copies so raw data remains unchanged
products_clean = products_check.copy()
orders_clean = orders_check.copy()
reviews_clean = reviews_check.copy()

# Step 2: Standardize column names
for df in [products_clean, orders_clean, reviews_clean]:
    df.columns = df.columns.str.strip().str.lower()

# Step 3: Remove duplicate records
products_clean = products_clean.drop_duplicates()
orders_clean = orders_clean.drop_duplicates()
reviews_clean = reviews_clean.drop_duplicates()

# Step 4: Remove rows with missing required values
products_clean = products_clean.dropna(
    subset=["product_id", "product_name", "category", "brand", "unit_price"]
)
orders_clean = orders_clean.dropna(
    subset=["order_id", "product_id", "order_date", "quantity",
            "unit_price", "order_status"]
)
reviews_clean = reviews_clean.dropna(
    subset=["review_id", "product_id", "review_date", "rating", "review_text"]
)

# Step 5: Standardize text fields
for col in ["product_name", "category", "brand"]:
    products_clean[col] = products_clean[col].astype(str).str.strip()

orders_clean["order_status"] = (
    orders_clean["order_status"].astype(str).str.strip().str.title()
)
reviews_clean["review_text"] = (
    reviews_clean["review_text"].astype(str).str.strip()
)

# Step 6: Convert columns to correct data types
products_clean["unit_price"] = pd.to_numeric(
    products_clean["unit_price"], errors="coerce"
)
orders_clean["quantity"] = pd.to_numeric(
    orders_clean["quantity"], errors="coerce"
)
orders_clean["unit_price"] = pd.to_numeric(
    orders_clean["unit_price"], errors="coerce"
)
reviews_clean["rating"] = pd.to_numeric(
    reviews_clean["rating"], errors="coerce"
)

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"], errors="coerce"
)
reviews_clean["review_date"] = pd.to_datetime(
    reviews_clean["review_date"], errors="coerce"
)

# Step 7: Apply basic business rules
products_clean = products_clean[
    products_clean["unit_price"] > 0
].copy()

orders_clean = orders_clean[
    (orders_clean["quantity"] > 0)
    & (orders_clean["unit_price"] > 0)
    & (orders_clean["order_status"].isin(
        ["Delivered", "Cancelled", "Returned"]
    ))
    & orders_clean["order_date"].notna()
].copy()

reviews_clean = reviews_clean[
    reviews_clean["rating"].between(1, 5)
    & reviews_clean["review_date"].notna()
    & reviews_clean["review_text"].ne("")
].copy()

# Step 8: Save cleaned files
CLEAN_DIR = PROJECT_DIR / "data" / "clean"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

products_clean.to_csv(CLEAN_DIR / "products_clean.csv", index=False)
orders_clean.to_csv(CLEAN_DIR / "orders_clean.csv", index=False)
reviews_clean.to_csv(CLEAN_DIR / "reviews_clean.csv", index=False)

# Step 9: Report results
print("CLEANING COMPLETE")
print("-" * 30)
print("Products:", len(products_clean))
print("Orders:", len(orders_clean))
print("Reviews:", len(reviews_clean))
print("Saved to:", CLEAN_DIR)
```
